```python
import pandas as pd
```

```python
df = pd.read_csv("../data/Food_Delivery_Time_Prediction.csv")
```

```python
str_cols = [
    "day_of_week",
    "weather",
    "pickup_zone",
    "dropoff_zone",
    "vehicle_type",
    "cuisine_type",
    "restaurant_load",
    "delivery_distance_category",
    "traffic_level",
    "delivery_priority"
]
```

```python
df.columns = df.columns.str.lower()
df["order_date"] = pd.to_datetime(df["order_date"])
df[str_cols] = df[str_cols].apply(lambda x : x.str.lower())
df["month"] = df["order_date"].dt.month_name()
```

```python
df.info()
```

```python
df["vehicle_type"].unique()
```

```python
traffic_lvl_map = {
    "low" : 1,
    "moderate" : 2,
    "high":3,
    "severe":4
}
```

```python
df["traffic_lvl_code"] = df["traffic_level"].map(traffic_lvl_map)
```

```python
df["rider_experience_group"] = pd.cut(
    df["rider_experience_years"],
    bins=[0, 2, 5, 8, 11, float("inf")],
    labels=["0–2 years", "3–5 years", "6–8 years", "9–11 years", "12+ years"]
)
```

```python
df["waiting_pct"] = (
    df["preparation_time_min"] / df["time_taken_min"] * 100
).round(2)
```

```python
df["riders_del_time"] = (
   df["time_taken_min"] -  df["preparation_time_min"] 
).round(2)
```
