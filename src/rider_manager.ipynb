```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet("../data/cleaned_data.parquet")
df.head(3)
```

```python
df["rider_experience_group"].unique()
```

```python
# Does more experience make riders faster? expeerience group to delivery time
exp_to_del_time = df.groupby("rider_experience_group").agg(
    avg_delivery_time=pd.NamedAgg(column="time_taken_min", aggfunc="mean")
).round(2).reset_index()
exp_to_del_time
```

```python
# Does rider rating predict delivery speed? 
rating_del_speed = df["rider_rating"].corr(df["time_taken_min"])
print(f"Correlation between rider rating and delivery speed: {rating_del_speed:.2f}")
```
